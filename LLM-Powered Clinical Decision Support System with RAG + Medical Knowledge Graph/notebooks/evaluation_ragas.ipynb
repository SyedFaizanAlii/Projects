## 5. Comprehensive Evaluation Dashboard
Generate summary statistics and evaluation dashboard across multiple clinical queries.

```python
class RAGASMetrics:
    """Metrics for Retrieval-Augmented Generation evaluation"""
    
    @staticmethod
    def faithfulness(answer: str, context: str) -> float:
        """
        Faithfulness: Measure how grounded the answer is in the context.
        Ranges from 0 (hallucinated) to 1 (fully grounded).
        """
        if not context or not answer:
            return 0.0
        
        # Simple implementation: word overlap ratio
        answer_words = set(answer.lower().split())
        context_words = set(context.lower().split())
        overlap = len(answer_words & context_words)
        max_words = max(len(answer_words), 1)
        
        return min(overlap / max_words, 1.0)
    
    @staticmethod
    def answer_relevancy(answer: str, query: str) -> float:
        """
        Answer Relevancy: Measure how relevant the answer is to the query.
        Ranges from 0 (irrelevant) to 1 (highly relevant).
        """
        if not answer or not query:
            return 0.0
        
        query_words = set(query.lower().split())
        answer_words = set(answer.lower().split())
        overlap = len(query_words & answer_words)
        
        return overlap / max(len(query_words), 1)
    
    @staticmethod
    def context_recall(context: str, ground_truth: str) -> float:
        """
        Context Recall: Measure how much relevant information is in the context.
        Ranges from 0 (no relevant context) to 1 (all relevant info present).
        """
        if not context or not ground_truth:
            return 0.0
        
        context_words = set(context.lower().split())
        truth_words = set(ground_truth.lower().split())
        overlap = len(context_words & truth_words)
        
        return overlap / max(len(truth_words), 1)
    
    @staticmethod
    def context_precision(context: str, query: str) -> float:
        """
        Context Precision: Measure absence of noise/irrelevance in context.
        Ranges from 0 (all noise) to 1 (all relevant).
        """
        if not context or not query:
            return 0.0
        
        query_words = set(query.lower().split())
        context_words = set(context.lower().split())
        relevant_words = query_words & context_words
        
        return len(relevant_words) / max(len(context_words), 1)

# Test RAGAS metrics
sample_query = "What are the interactions between metformin and lisinopril?"
sample_context = "Metformin is a first-line medication for diabetes. Lisinopril is an ACE inhibitor. Drug interaction alert: Metformin and ACE inhibitors may increase lactic acidosis risk."
sample_answer = "Metformin and ACE inhibitors like lisinopril together may increase risk of lactic acidosis in renal insufficiency."

metrics = RAGASMetrics()
results = {
    "Faithfulness": metrics.faithfulness(sample_answer, sample_context),
    "Answer Relevancy": metrics.answer_relevancy(sample_answer, sample_query),
    "Context Recall": metrics.context_recall(sample_context, sample_query),
    "Context Precision": metrics.context_precision(sample_context, sample_query),
}

print("RAGAS Evaluation Metrics:")
for metric_name, score in results.items():
    print(f"  {metric_name}: {score:.3f}")

# Visualization
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Bar chart
metrics_list = list(results.keys())
scores = list(results.values())
ax1.bar(metrics_list, scores, color=['#3b82f6', '#10b981', '#f59e0b', '#ef4444'])
ax1.set_ylim([0, 1.0])
ax1.set_title('RAGAS Metrics Scores')
ax1.set_ylabel('Score')
ax1.grid(axis='y', alpha=0.3)

# Summary
summary_text = f"""
RAGAS Evaluation Summary
========================
Query: {sample_query[:50]}...

Average Score: {np.mean(scores):.3f}
Min Score: {min(scores):.3f}
Max Score: {max(scores):.3f}

Overall Quality: {'Excellent' if np.mean(scores) > 0.8 else 'Good' if np.mean(scores) > 0.6 else 'Fair'}
"""
ax2.text(0.1, 0.5, summary_text, fontsize=10, family='monospace', verticalalignment='center')
ax2.axis('off')

plt.tight_layout()
plt.show()
```
